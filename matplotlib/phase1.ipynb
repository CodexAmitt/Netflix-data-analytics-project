```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
a = np.linspace(0,5,11)
a
b = a**2 
b
```

```python
plt.plot(a,b)
plt.title("H Na Mast Graph")
plt.xlabel("a axis")
plt.ylabel("b axis")
```

```python
plt.subplot(2,2,2)
plt.plot(a**2,a**2)
plt.subplot(2,2,1)
plt.plot(a,b)
plt.subplot(2,2,3)
plt.plot(a,b)
plt.subplot(2,2,4)
plt.plot(b,a)

```

## Object oriented way

```python
fig = plt.figure(figsize=(10,8),dpi=100)
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(a,b)
axis2 = fig.add_axes([0.2,0.4,0.4,0.4])
axis2.plot(a,b)
```

## Types of plot

```python
plt.scatter(a,b)
```

```python
plt.hist(a)
```

```python
plt.boxplot(a)
```

## Saving plots

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(a,b)
plt.savefig("basicplot.png")
```
