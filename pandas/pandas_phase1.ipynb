## Creating data frame

```python
data = {
    'name' : ['john','anna','peter','linda'],
    'age': [28,34,29,42],
    'city': ['new york','paris','berlin','london'],
    'salary': [65000,70000,62000,85000]

}
pd.DataFrame(data)
```

```python
data_list = [
    ['john',28,	'new york',	65000],
	['anna',34,	'paris',	70000],
	['peter',29,'berlin',	62000],
	['linda',42,'london',	85000]

]
df2 = pd.DataFrame(data_list)
columns = ['name','age','city','salary']
df2 = pd.DataFrame(data_list,columns=columns)
df2
```

```python
## select column
#df2['name']

df2[['name','city']]
```

```python
#add new column
df2["designation"] = ['doc','eng','doc','eng']
df2
```

```python
#remove a column
df2.drop(['city','salary'],axis=1)
```

```python
# remove column completely
df2.drop('designation',axis=1,inplace = True)
```

```python
#remove row
df2.drop(1,axis= 0)
```

## Selecting a row

```python
#for single selection
df2.loc[0]
```

```python
#for multiple selection
df2.loc[[0,1]]
```

```python
# for select indexwise
df2.iloc[3]
```

## selecting subset of rows and columns

```python
df2.loc[[0,1]][['city','salary']]
```

```python
df2.loc[[2,3]][['name','age']]
```

## conditional selection

```python
# i only want to see those people whose age is above 30
df2[df2['age'] > 30]
```

```python
#i only want people whose age is above 30 and their city must be paris
df2[(df2['age'] > 30) & (df2['city'] == 'paris')]
```
