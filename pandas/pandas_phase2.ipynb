## Missing data


```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1,2,np.nan,4,5],
    'B': [1,2,3,4,5],
    'C': [1,2,3,np.nan,np.nan],
    'D': [1,np.nan,np.nan,np.nan,5]
}
df = pd.DataFrame(data)
```

```python
#check null value
df.isna()
```

```python
#count how many null value are there
df.isna().sum()
```

```python
#check column wise null available or not // searching
df.isna().any()
```

```python
#remove row of null values
df.dropna()
```

```python
#print condition wise null value
df.dropna(thresh=3)
```

```python
#filling the missing data
df.fillna(0)
```

```python
#fill the different value for each column
values = {'A':0,'B':100,'C':300,'D':400}
df.fillna(value=values)
```

```python
df.fillna(df.mean())
```

# merging joining and concatination

```python
#merging 2 data frames
emp = pd.DataFrame({
    'emp_id': [1,2,3,4,5],
    'name': ['john','Anna','peter','linda','bob'],
    'department': ['HR','IT','Finance','IT','HR']
})

#Dataframe 2: salary information
salary = pd.DataFrame({
    'emp_id': [1,2,3,6,7],
    'salary': [60000,80000,65000,70000,90000],
    'bonus': [5000,10000,7000,8000,12000]
})
```

```python
emp
```

```python
salary
```

```python
#merge id with salary
pd.merge(emp,salary,on='emp_id',how='inner')
```

```python
pd.merge(emp,salary,on='emp_id',how='outer')
```

```python
pd.merge(emp,salary,on='emp_id',how='left')
```

```python
pd.merge(emp,salary,on='emp_id',how='right')
```

```python
#concatination of data frame
df1 = pd.DataFrame({
    'A': ['A0','A1','A2'],
    'B': ['B0','B2','B3'],
    'C': ['C0','C1','C2']
})

df2 = pd.DataFrame({
    'A': ['A3','A4','A5'],
    'B': ['B3','B4','B5'],
    'C': ['C3','C3','C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat((df1,df2))
```

```python
pd.concat((df1,df2),axis=1)
```

```python
#joining 2 data frames
df1 = pd.DataFrame({
    'name': ['alice','bob','charlie'],
}, index=[1,2,3])

df2= pd.DataFrame({
    'score': [85,90,75]
}, index=[2,3,4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2,how='outer')
```
