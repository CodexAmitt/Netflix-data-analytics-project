```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 150000], #paradise
    [2, 120000, 140000, 160000, 190000], #Beijing bites
    [3, 200000, 230000, 260000, 300000], #pizza hub
    [4, 180000, 210000, 240000, 270000], #burger point
    [5, 160000, 185000, 205000, 230000], #chai point
])

print("=== Zomato sales analysis ===")
print("\n sales data shape", sales_data.shape)
print("\n sample data for 1st 3 restaurant \n", sales_data[:3])
```

```python
# total shale per year
print(np.sum(sales_data, axis=0))
yearly_total = np.sum(sales_data[:,1:], axis=0)
print(yearly_total)

#minimum sale per restaurant
min_sale = np.min(sales_data[:,1:],axis=1)
print(min_sale)

#maximum sale per year
max_sale = np.max(sales_data[:,1:], axis=0)
print(max_sale)

#Average sale per year
avg_sale = np.mean(sales_data[:,1:],axis=1)
print(avg_sale)

#cumulative sum
cumsum = np.cumsum(sales_data[:,1:],axis=1)
print("cumsum is",cumsum)

plt.figure(figsize=(8,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales accross all restaurant")
plt.xlabel("years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```
